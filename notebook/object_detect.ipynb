```python
import torch
from PIL import Image, ImageDraw
from torchvision.transforms import functional as F

# Load the DETR model
model = torch.hub.load('facebookresearch/detr', 'detr_resnet50', pretrained=True)
model.eval()

# Load and preprocess the image
image = Image.open('img_test.jpg')
image_tensor = F.to_tensor(image).unsqueeze(0)  # Add batch dimension

# Perform object detection
with torch.no_grad():
    outputs = model(image_tensor)

# Post-process the outputs
# Get the predicted boxes and labels
probas = outputs['pred_logits'].softmax(-1)  # Apply softmax to get probabilities
keep = probas.max(-1).values > 0.9  # Keep only predictions with a threshold of 0.9
boxes = outputs['pred_boxes'][keep].cpu().numpy()  # Get the boxes
labels = probas[keep].cpu().numpy()  # Get the labels

# Draw the bounding boxes on the image
draw = ImageDraw.Draw(image)
for box, label in zip(boxes, labels):
    x_min, y_min, x_max, y_max = box
    draw.rectangle([x_min, y_min, x_max, y_max], outline="red", width=2)
    draw.text((x_min, y_min), f"{label.argmax()}", fill="red")

# Save or display the output image
image.show()  # To display the image
image.save('output_image.jpg')  # To save the image
```

```python
import torch
from PIL import Image, ImageDraw
from torchvision.transforms import functional as F

# Load the DETR model
model = torch.hub.load('facebookresearch/detr', 'detr_resnet50', pretrained=True)
model.eval()

# Load and preprocess the image
image_path = 'img_test_2.jpg'  # Change this to your image path
image = Image.open(image_path)
image_tensor = F.to_tensor(image).unsqueeze(0)  # Add batch dimension

# Perform object detection
with torch.no_grad():
    outputs = model(image_tensor)

# Post-process the outputs
# Get the predicted boxes and labels
probas = outputs['pred_logits'].softmax(-1)  # Apply softmax to get probabilities
keep = probas.max(-1).values > 0.9  # Keep only predictions with a threshold of 0.9
boxes = outputs['pred_boxes'][keep].cpu().numpy()  # Get the boxes

# Crop and save each detected object
for i, box in enumerate(boxes):
    x_min, y_min, x_max, y_max = box

    # Ensure coordinates are within image dimensions
    x_min = max(0, min(x_min, image.width))
    y_min = max(0, min(y_min, image.height))
    x_max = max(0, min(x_max, image.width))
    y_max = max(0, min(y_max, image.height))

    # Only crop if the coordinates are valid
    if x_min < x_max and y_min < y_max:
        cropped_image = image.crop((x_min, y_min, x_max, y_max))
        cropped_image.save(f'cropped_object_{i}.jpg')  # Save the cropped image
```
